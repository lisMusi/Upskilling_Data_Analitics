```python
import matplotlib.pyplot as plt
import numpy as np

eu_countries = [
    "Austria", "Belgium", "Bulgaria", "Croatia", "Cyprus", "Czechia",
    "Denmark", "Estonia", "Finland", "France", "Germany", "Greece",
    "Hungary", "Ireland", "Italy", "Latvia", "Lithuania", "Luxembourg",
    "Malta", "Netherlands", "Poland", "Portugal", "Romania", "Slovak Republic",
    "Slovenia", "Spain", "Sweden", "United Kingdom"
]

europe_data = df_nac_mundo[df_nac_mundo['Country Name'].isin(eu_countries)]
years = list(map(str, range(1960, 2024)))
data = europe_data.set_index('Country Name')[years]  # cada fila representa un país, cada columna un año
data = data.fillna(0)

highlight_category = 'Spain'

plt.figure(figsize=(22, 10))
bottom = np.zeros(len(years))  # inicializar en 0

for country in data.index:
    if country == highlight_category:
        plt.bar(years, data.loc[country], bottom=bottom, label=country, alpha=1.0)
    else:
        plt.bar(years, data.loc[country], bottom=bottom, label=country, alpha=0.3)
    bottom += data.loc[country]

plt.xlabel('Año')
plt.ylabel('Nacimientos por cada 1000 habitantes')
plt.title('Nacimientos en países de la UE por cada 1000 habitantes')
plt.xticks(rotation=90)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize='small')
plt.tight_layout()
plt.show()
```

```python
# Nacimientos x distritos:

import pandas as pd

df_nac_distritos = pd.read_csv('/Users/liss/Desktop/Proyecto_Final/Naixements_districtes.csv', sep=',')
print(df_nac_distritos.head())
```

```python
import pandas as pd

poblacion_distritos = pd.read_csv('/Users/liss/Desktop/Proyecto_Final/Poblacion_distritos.csv', sep=',', on_bad_lines='skip')
poblacion_distritos.head()
```

```python
df_pobl_distritos = poblacion_distritos[poblacion_distritos['Tipo de territorio'] == 'Districte']
df_pobl_distritos.head()
```

```python
# Convertir el DataFrame a formato largo (long format)
df_long = df_pobl_distritos.melt(
    id_vars=['Territorio', 'Tipo de territorio'], 
    value_vars=[str(year) for year in range(1997, 2024)],  
    var_name='Año', 
    value_name='Población'
)

# Ver cómo quedó el DataFrame
print(df_long.head())
```

```python
df_long.rename(columns={'Territorio': 'Territori', 'Tipo de territorio': 'Tipus de territori', 'Año': 'Any'}, inplace=True)
df_long.head()
```

```python
df_nac_distritos['Any'] = df_nac_distritos['Any'].astype(int)
df_long['Any'] = df_long['Any'].astype(int)
```

```python
df_distritos = df_nac_distritos.merge(df_long[['Territori', 'Tipus de territori', 'Any', 'Población']],
                                      on=['Territori', 'Tipus de territori', 'Any'], how='left')
df_distritos
```

```python
df_distritos['Tasa_de_Natalidad'] = (df_distritos['Naixements'] / df_distritos['Población']) * 1000
df_distritos.head()
```

```python
import matplotlib.pyplot as plt

tasa_media_distritos = df_distritos.groupby('Territori')['Tasa_de_Natalidad'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
tasa_media_distritos.plot(kind='bar', color='skyblue')
plt.title('Tasa media de natalidad por distritos (por cada 1000 habitantes)')
plt.xlabel('Distrito')
plt.ylabel('Tasa de natalidad')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.grid(axis='y')
plt.show()
```

```python
import pandas as pd
import plotly.express as px

fig = px.scatter_mapbox(
    df_distritos,
    lat="Latitut",
    lon="Longitut",
    color="Territori", 
    size="Tasa_de_Natalidad",
    animation_frame="Any",
    color_continuous_scale=px.colors.sequential.Plasma,
    size_max=15,
    zoom=11,
    mapbox_style="carto-positron",
    title="Nacimientos en Barcelona por distritos"
)

fig.write_html("nacimientos_distritos_mapa.html")
fig.show()
```

# Correlación entre nacimientos, mascotas y consumo de carne:

```python
# Seleccionar las columnas relevantes
df_cat = df_cat[['Tasa_de_Natalidad', 'Tasa_de_Mascotas', 'Consumo_Carne_x_Capita']]

# Eliminar filas con NaN en cualquiera de las columnas
df_cat = df_cat.dropna()
```

```python
# HISTOGRAMA Y KDE:

import matplotlib.pyplot as plt
import seaborn as sns


fig, axes = plt.subplots(1, 3, figsize=(12, 4))  

sns.histplot(df_cat['Tasa_de_Natalidad'], kde=True, ax=axes[0])
axes[0].set_title('Tasa_de_Natalidad')

sns.histplot(df_cat['Tasa_de_Mascotas'], kde=True, ax=axes[1])
axes[1].set_title('Tasa_de_Mascotas')

sns.histplot(df_cat['Consumo_Carne_x_Capita'], kde=True, ax=axes[2])
axes[2].set_title('Consumo_Carne_x_Capita')

plt.tight_layout() 
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import scipy.stats as stats

# PRUEBAS DE NORMALIDAD
variables = ['Tasa_de_Natalidad', 'Tasa_de_Mascotas', 'Consumo_Carne_x_Capita']
for var in variables:
    stat, p = stats.shapiro(df_cat[var])
    print(f"{var} - Shapiro-Wilk test: Estadístico={stat:.3f}, p-valor={p:.3f}")

# GRÁFICOS Q-Q
import statsmodels.api as sm
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for i, var in enumerate(variables):
    sm.qqplot(df_cat[var], line='s', ax=axes[i])
    axes[i].set_title(f'Q-Q plot: {var}')
plt.tight_layout()
plt.show()
```

```python
# CORRELACIÓN DE PEARSON:

import seaborn as sns
import matplotlib.pyplot as plt

# Calcular la matriz de correlación entre las variables
correlation_matrix = df_cat.corr()
print(correlation_matrix)

# Crear un mapa de calor para la matriz de correlación
plt.figure(figsize=(6, 5))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1,
            xticklabels=['Natalidad', 'Mascotas', 'Consumo Carne'],
            yticklabels=['Natalidad', 'Mascotas', 'Consumo Carne'])
            # cambiamos los nombres de las variables solo para la visualización
plt.title('Correlación Pearson')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# CORRELACIÓN SPEARMAN Y KENDALL
corr_spearman = df_cat[variables].corr(method='spearman')
corr_kendall = df_cat[variables].corr(method='kendall')

# MAPAS DE CALOR
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.heatmap(corr_spearman, annot=True, cmap='coolwarm', vmin=-1, vmax=1, xticklabels=['Natalidad', 'Mascotas', 'Consumo Carne'],
            yticklabels=['Natalidad', 'Mascotas', 'Consumo Carne'], ax=axes[0])
axes[0].set_title("Correlación Spearman")

sns.heatmap(corr_kendall, annot=True, cmap='coolwarm', vmin=-1, vmax=1,  xticklabels=['Natalidad', 'Mascotas', 'Consumo Carne'],
            yticklabels=['Natalidad', 'Mascotas', 'Consumo Carne'], ax=axes[1])
axes[1].set_title("Correlación Kendall")

plt.tight_layout()
plt.show()
```

```python
# Detección de outliers:

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# BOXPLOTS PARA DETECCIÓN DE OUTLIERS
plt.figure(figsize=(12, 4))
for i, var in enumerate(variables):
    plt.subplot(1, 3, i+1)
    sns.boxplot(y=df_cat[var], color='skyblue')
    plt.title(f'Boxplot: {var}')
plt.tight_layout()
plt.show()
```

```python
# PRUEBAS DE SIGNIFICANCIA PARA LA CORRELACIÓN 

import pandas as pd
from scipy.stats import pearsonr, spearmanr, kendalltau

vars = ['Tasa_de_Natalidad', 'Tasa_de_Mascotas', 'Consumo_Carne_x_Capita']
combinaciones = [('Tasa_de_Natalidad', 'Tasa_de_Mascotas'),
                 ('Tasa_de_Natalidad', 'Consumo_Carne_x_Capita'),
                 ('Tasa_de_Mascotas', 'Consumo_Carne_x_Capita')]

print("PRUEBAS DE SIGNIFICANCIA PARA CORRELACIONES:")
for x, y in combinaciones:
    pearson_r, p_p = pearsonr(df_cat[x], df_cat[y])
    spearman_r, p_s = spearmanr(df_cat[x], df_cat[y])
    kendall_r, p_k = kendalltau(df_cat[x], df_cat[y])
    
    print(f"\nVariables: {x} vs {y}")
    print(f"  Pearson:  r = {pearson_r:.2f}, p-valor = {p_p:.4f}")
    print(f"  Spearman: r = {spearman_r:.2f}, p-valor = {p_s:.4f}")
    print(f"  Kendall:  r = {kendall_r:.2f}, p-valor = {p_k:.4f}")
```

```python
# REGRESIÓN LINEAL:

import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Primer gráfico
r1, _ = pearsonr(df_cat["Tasa_de_Natalidad"], df_cat["Tasa_de_Mascotas"])
sns.regplot(
    x=df_cat["Tasa_de_Natalidad"], y=df_cat["Tasa_de_Mascotas"],
    line_kws={"color": "black", "linewidth": 1.5, "alpha": 0.5},
    color="b", ax=axes[0]
)
axes[0].set_title("Natalidad vs Mascotas")
axes[0].text(0.05, 0.95, f'R² = {r1**2:.2f}', transform=axes[0].transAxes,
             fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle="round", facecolor="white", alpha=0.5))

# Segundo gráfico
r2, _ = pearsonr(df_cat["Tasa_de_Natalidad"], df_cat["Consumo_Carne_x_Capita"])
sns.regplot(
    x=df_cat["Tasa_de_Natalidad"], y=df_cat["Consumo_Carne_x_Capita"],
    line_kws={"color": "black", "linewidth": 1.5, "alpha": 0.5},
    color="b", ax=axes[1]
)
axes[1].set_title("Natalidad vs Consumo de Carne")
axes[1].text(0.05, 0.95, f'R² = {r2**2:.2f}', transform=axes[1].transAxes,
             fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle="round", facecolor="white", alpha=0.5))

# Tercer gráfico
r3, _ = pearsonr(df_cat["Tasa_de_Mascotas"], df_cat["Consumo_Carne_x_Capita"])
sns.regplot(
    x=df_cat["Tasa_de_Mascotas"], y=df_cat["Consumo_Carne_x_Capita"],
    line_kws={"color": "black", "linewidth": 1.5, "alpha": 0.5},
    color="b", ax=axes[2]
)
axes[2].set_title("Mascotas vs Consumo de Carne")
axes[2].text(0.05, 0.95, f'R² = {r3**2:.2f}', transform=axes[2].transAxes,
             fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle="round", facecolor="white", alpha=0.5))

plt.tight_layout()
plt.show()
```

- R² < 0.3: relación muy débil.
- 0.3 ≤ R² < 0.5: relación moderada.
- 0.5 ≤ R² < 0.7: relación fuerte.
- R² ≥ 0.7: relación muy fuerte.
