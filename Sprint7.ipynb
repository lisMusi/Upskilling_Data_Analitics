### Exercici 1 
Calculadora de l'índex de massa corporal:

- Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. 

- La funció ha de classificar el resultat en les seves respectives categories.

```python
# Creo una función para obtener los datos que necesito:

def obtener_datos():
    peso = float(input("Introduzca su peso en kg: "))
    altura = float(input("Introduzca su altura en centimetros: "))

    return peso, altura


# Creo otra función para calcular el IMC y clasificarlo:

def calcular_IMC(peso, altura):
    IMC = peso / ((altura*0.01)** 2)

    if IMC <= 18.5:
        clasificacion = "BAJOPESO"
    elif 18.5 < IMC <= 24.9:
        clasificacion = "NORMOPESO"
    elif 25 <= IMC <= 29.9:
        clasificacion = "SOBREPESO"
    else:
        clasificacion = "OBESIDAD"

    return IMC, clasificacion


# Por último utilizo otra función para imprimir el resultado:

def imprimir_resultado_IMC(IMC, clasificacion):
    print(f"Tu IMC es: {IMC:.2f} y tu clasificación es {clasificacion}")


# Utilizo la función main() para organizar la ejecución del código:

def main():
    peso, altura = obtener_datos()
    IMC, clasificacion = calcular_IMC(peso, altura)
    imprimir_resultado_IMC(IMC, clasificacion)
main()
```

### Exercici 2
Convertidor de temperatures:

- Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions.

```python
# Creo una función para obtener la temperatura:

def obtener_temperatura():
    temperatura = input("Escribe la unidad en la que está tu temperatura: C (para grados celsius), F (para fahrenheit) o K (para kelvin)").strip().lower() # para evitar problemas con espacios o mayúsculas
    if temperatura == 'c':
        valor = float(input("Introduzca la temperatura en grados celsius (solo el número): "))
    elif temperatura == 'f':
        valor = float(input("Introduzca la temperatura en grados fahrenheit (solo el número): "))
    elif temperatura == 'k':
        valor = float(input("Introduzca la temperatura en kelvins (solo el número): "))
    else:
        print("Unidad no válida!")
        return None, None  # retorna None si la unidad no es válida
    
    return valor, temperatura

# Creo otra función para calcular la temperatura:

def convertidor_temperatura(valor, temperatura):
    if temperatura == 'c':
        fahrenheit = (valor * 9/5) + 32
        kelvin = valor + 273.15
        rankine = (valor + 273.15) * 9/5
        reaumur = valor * 0.8
        print(f"{valor}ºC equivalen a {fahrenheit:.2f}ºF, {kelvin:.2f}K, {rankine:.2f}ºRa y {reaumur:.2f}ºRe")
    
    elif temperatura == 'f':
        celsius = (valor - 32) * 5/9
        kelvin = (valor + 459.67) * 5/9
        rankine = valor + 459.67
        reaumur = (valor - 32) * 4/9 
        print(f"{valor}ºF equivalen a {celsius:.2f}ºC, {kelvin:.2f}K, {rankine:.2f}ºRa y {reaumur:.2f}ºRe")

    elif temperatura == 'k':
        celsius = valor - 273.15
        fahrenheit = (valor * 9/5) - 459.67
        rankine = valor * 9/5
        reaumur = (valor - 273.15) / 1.25
        print(f"{valor}K equivalen a {celsius:.2f}ºC, {fahrenheit:.2f}ºF, {rankine:.2f}ºRa y {reaumur:.2f}ºRe")


# Ejecutar el código
valor, temperatura = obtener_temperatura()
if valor is not None and temperatura is not None:
    convertidor_temperatura(valor, temperatura)
```

### Exercici 3
Comptador de paraules d'un text:

- Escriu una funció que donat un text, mostri les vegades que apareix cada paraula.

```python
import re # importo el módulo re primero para poder utilizar la funcion re.sub(). 
# re es la librería estándar para trabajar con expresiones regulares (regex): permite buscar, reemplazar, dividir y manipular cadenas de texto de manera avanzada.

# Creo una función para contar las palabras de un texto

def contador_palabras(texto):
    texto = texto.lower()  # convertimos el texto a minúsculas
    texto = re.sub(r'[^\w\s]', '', texto)  # eliminamos los signos de puntuación

    palabras = {}  
    lista_palabras = texto.split()  # dividimos el texto en palabras

    for palabra in lista_palabras:
        palabras[palabra] = palabras.get(palabra, 0) + 1  # usamos .get() para acceder de forma segura a los valores del diccionario sin generar errores si la clave no existe
    
    palabras = dict(sorted(palabras.items())) # con sorted ordeno el diccionario en orden alfabético de la key

    return palabras  


# Creo una función para ejecutar la función anterior:

def ejecutar_contador_palabras():
    texto = input("Introduzca un texto: ")
    resultado = contador_palabras(texto)
    print(resultado)


# Llamo a la función 'ejecutar' para ejecutarlo todo:

ejecutar_contador_palabras()
```

### Exercici 4
Diccionari invers:

- Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviats les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment.

```python
# Creo una función que me devuelva el diccionario inverso

def diccionario_inverso(diccionario):
    diccionario_nuevo = {}

    for key, value in diccionario.items(): 
        if value in diccionario_nuevo:
            raise ValueError(f"Error: múltiples keys para el value '{value}'")  # lanza un error, raise se utiliza para lanzar una excepción intencionalmente
        diccionario_nuevo[value] = key # cambia el orden clave:valor

    return diccionario_nuevo

diccionario = {'x': 'apple', 'y': 'banana', 'z': 'banana'} # diccionario de prueba

# probamos que imprime error en valores duplicados:
try: # usamos try-except para que nos muestre el error si hay valores duplicados
    diccionario_nuevo = diccionario_inverso(diccionario)
    print(diccionario_nuevo)
except ValueError as e:
    print(e)
```

```python
diccionario2 = {'x': 'apple', 'y': 'banana', 'z': 'strawberry'}

# probamos que funciona si no hay valores duplicados:
try:
    diccionario_nuevo = diccionario_inverso(diccionario2)
    print(diccionario_nuevo)
except ValueError as e:
    print(e)
```

### Exercici 1
Diccionari invers amb duplicats:

- Continuant amb l'exercici 4 del nivell 1: al client es va oblidar de comentar un detall i resulta que els valors en el diccionari original poden duplicar-se i més, per la qual cosa les claus intercanviades poden tenir duplicats. En aquest cas, en l'exercici anterior imprimies un missatge d'advertiment, ara, els valors del diccionari resultant hauran d'emmagatzemar-se com una llista. Tingues en compte que si és un valor únic no ha de ser una llista.

```python
# Creo una nueva función para modificar la funcion diccionario_inverso y almacenar los valores duplicados en una lista 

def valores_duplicados(diccionario): # llama a diccionario_inverso y conviete los valores duplicados en lista
    try:
        return diccionario_inverso(diccionario)  # intenta invertir el diccionario normalmente
    except ValueError:
        diccionario_nuevo = {}  # creo un nuevo diccionario para manejar duplicados

        for key, value in diccionario.items():
            if value in diccionario_nuevo: # si el value ya existe, lo convertimos en lista y agregamos la nueva key
                if isinstance(diccionario_nuevo[value], list): # la función isinstance(objeto, tipo) verifica si un objeto es de un determinado tipo 
                    diccionario_nuevo[value].append(key)
                else:
                    diccionario_nuevo[value] = [diccionario_nuevo[value], key]
            else:
                diccionario_nuevo[value] = key  # si no existe, lo guardamos como string 

        return diccionario_nuevo


diccionario = {'x': 'apple', 'y': 'banana', 'z': 'banana'}

diccionario_nuevo = valores_duplicados(diccionario)
print(diccionario_nuevo)
```

### Exercici 2
Conversió de tipus de dades:

- El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']].

```python
# Creo una función que aplana la lista (eliminando las listas y tuplas en su interior) y clasifica los elementos en números decimales u otros

def procesar_lista(lista):
    lista_flotantes = []
    lista_otros = []
    
    def aplanar_y_organizar(elemento): # aplana la lista y clasifica los elementos
        if isinstance(elemento, (list, tuple)):  # la función isinstance(objeto, tipo) verifica si un objeto es de un determinado tipo 
            for sub_elemento in elemento:
                aplanar_y_organizar(sub_elemento) # si es una lista o tupla, la procesamos recursivamente
        else:
            try: # usamos try-except para evitar que el programa falle con valores no numéricos
                lista_flotantes.append(float(elemento))  # intentamos convertir a float, si funciona: es un número decimal (flotante)
            except (ValueError, TypeError):  # valor incorrecto o tipo de dato incorrecto para la operación 
                lista_otros.append(elemento)  # si no se puede convertir, lo agregamos a otra lista

    for elemento in lista:  # procesamos la lista principal
        aplanar_y_organizar(elemento)
    
    return lista_flotantes, lista_otros

lista = ['1.3', 'one', '1e10', 'seven', '3-1/2', ('2', 1, 1.4, 'not-a-number'), [1, 2, '3', '3.4']]

lista_flotantes, lista_otros = procesar_lista(lista) # llamar a la función

print(f"lista_flotantes: {lista_flotantes}")
print(f"lista_otros: {lista_otros}")
```

### Exercici 1
Comptador i endreçador de paraules d'un text:

- El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. 

```python
import re  # importo el módulo re primero para poder utilizar la funcion re.sub()

def contador_texto_ordenado(nombre_archivo):
    # Lectura del archivo de texto
    with open(nombre_archivo, "r", encoding="utf-8") as archivo:
        texto = archivo.read()  # leer el contenido del archivo
    
    # Preprocesamiento del texto antes de contarlo
    texto = texto.lower()  # convertimos a minúsculas
    texto = re.sub(r'[^\w\s]', '', texto)  # eliminamos signos de puntuación
    texto = texto.replace("\n", " ")  # eliminamos saltos de línea

    dicc_texto = {}  
    lista_texto = sorted(texto.split())  # dividimos en palabras y ordenamos

    for palabra in lista_texto:
        letra_inicial = palabra[0]  # obtener la primera letra

        if letra_inicial not in dicc_texto:
            dicc_texto[letra_inicial] = {}  # crear la key si no existe

        if palabra in dicc_texto[letra_inicial]:
            dicc_texto[letra_inicial][palabra] += 1  # si la palabra ya está: aumentar el contador
        else:
            dicc_texto[letra_inicial][palabra] = 1  # si la palabra no está: inicializar el contador

    return dicc_texto

ruta_archivo = "/Users/liss/Desktop/ItAcademy/2-Especialización/¡Sprints!/S7/Datos/tu_me_quieres_blanca.txt" # ruta del archivo de texto

dicc_texto_ordenado = contador_texto_ordenado(ruta_archivo) # llamamos a la función con el nombre del archivo

dicc_texto_ordenado # visualizamos el resultado
```
