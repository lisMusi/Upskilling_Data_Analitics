### Exercici 1 
Calculadora de l'índex de massa corporal:

- Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. 

- La funció ha de classificar el resultat en les seves respectives categories.

```python
def calcular_IMC(peso, altura):
    IMC = peso / (altura ** 2)

    if IMC <= 18.5:
        clasificacion = "BAJOPESO"
    elif 18.5 < IMC <= 24.9:
        clasificacion = "NORMOPESO"
    elif 25 <= IMC <= 29.9:
        clasificacion = "SOBREPESO"
    else:
        clasificacion = "OBESIDAD"

    return IMC, clasificacion

peso = float(input("Introduzca su peso en kg: "))
altura = float(input("Introduzca su altura en metros: "))

IMC, clasificacion = calcular_IMC(peso, altura)

print(f"Tu IMC es: {IMC:.2f} y tu clasificación es {clasificacion}")
```

### Exercici 2
Convertidor de temperatures:

- Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions.

```python
def convertidor_temperatura(celsius):
    fahrenheit = (celsius * 9/5) + 32
    kelvin = celsius + 273.15
    rankine = (celsius + 273.15) * 9/5
    reaumur = celsius * 0.8

    return fahrenheit, kelvin, rankine, reaumur

celsius = float(input("Introduzca la temperatura en grados celsius: "))

fahrenheit, kelvin, rankine, reaumur = convertidor_temperatura(celsius)

print(f"{celsius}ºC equivalen a {fahrenheit:.2f}ºF, {kelvin:.2f}K, {rankine:.2f}ºRa y {reaumur:.2f}ºRe")
```

### Exercici 3
Comptador de paraules d'un text:

- Escriu una funció que donat un text, mostri les vegades que apareix cada paraula.

```python
import re # importo el módulo re primero para poder utilizar la funcion re.sub(). 
# re es la librería estándar para trabajar con expresiones regulares (regex): permite buscar, reemplazar, dividir y manipular cadenas de texto de manera avanzada.

def contador_palabras(texto):
    texto = texto.lower()  # convertimos el texto a minúsculas
    texto = re.sub(r'[^\w\s]', '', texto)  # eliminamos los signos de puntuación pero no los espacios
    
    palabras = {} 
    lista_palabras = texto.split() # dividimos el texto en palabras

    for palabra in lista_palabras:
        if palabra in palabras:
            palabras[palabra] += 1 
        else:
            palabras[palabra] = 1  

    return palabras 

texto = input("Introduzca un texto: ")

palabras = contador_palabras(texto)

print(palabras)
```

### Exercici 4
Diccionari invers:

- Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviats les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment.

```python
def diccionario_inverso(diccionario):
    diccionario_nuevo = {}

    for key, value in diccionario.items(): 
        if value in diccionario_nuevo:
            raise ValueError(f"Error: múltiples keys para el value '{value}'")  # lanza un error, raise se utiliza para lanzar una excepción intencionalmente
        diccionario_nuevo[value] = key # cambia el orden clave:valor

    return diccionario_nuevo

diccionario = {'x': 'apple', 'y': 'banana', 'z': 'banana'} 

# probamos que imprime error en valores duplicados:
try: # usamos try-except para que nos muestre el error si hay valores duplicados
    diccionario_nuevo = diccionario_inverso(diccionario)
    print(diccionario_nuevo)
except ValueError as e:
    print(e)
```

```python
diccionario2 = {'x': 'apple', 'y': 'banana', 'z': 'strawberry'}

# probamos que funciona si no hay valores duplicados:
try:
    diccionario_nuevo = diccionario_inverso(diccionario2)
    print(diccionario_nuevo)
except ValueError as e:
    print(e)
```

### Exercici 1
Diccionari invers amb duplicats:

- Continuant amb l'exercici 4 del nivell 1: al client es va oblidar de comentar un detall i resulta que els valors en el diccionari original poden duplicar-se i més, per la qual cosa les claus intercanviades poden tenir duplicats. En aquest cas, en l'exercici anterior imprimies un missatge d'advertiment, ara, els valors del diccionari resultant hauran d'emmagatzemar-se com una llista. Tingues en compte que si és un valor únic no ha de ser una llista.

```python
def diccionario_inverso2(diccionario):
    diccionario_nuevo2 = {}

    for key, value in diccionario.items(): 
        if value not in diccionario_nuevo2:
            diccionario_nuevo2[value] = key  # si es el primer valor, lo guardamos como string
        else:
            if isinstance(diccionario_nuevo2[value], list): # la función isinstance(objeto, tipo) verifica si un objeto es de un determinado tipo 
                diccionario_nuevo2[value].append(key)  # si ya es una lista, agregamos
            else:
                diccionario_nuevo2[value] = [diccionario_nuevo2[value], key]  # si no, convertimos en lista

    return diccionario_nuevo2

diccionario = {'x': 'apple', 'y': 'banana', 'z': 'banana'} 

diccionario_nuevo2 = diccionario_inverso2(diccionario)

print(diccionario_nuevo2)
```

### Exercici 2
Conversió de tipus de dades:

- El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']].

```python
# Primero añado una función para aplanar la lista

def aplanar(lista):
    lista_plana = []
    for elemento in lista:
        if isinstance(elemento, (list, tuple)): # la función isinstance(objeto, tipo) verifica si un objeto es de un determinado tipo 
            lista_plana.extend(aplanar(elemento)) # si es una lista o una tupla, la procesamos recursivamente
        else:
            lista_plana.append(elemento)  # si no, lo agregamos tal cual
    return lista_plana

lista = ['1.3', 'one', '1e10', 'seven', '3-1/2', ('2', 1, 1.4, 'not-a-number'), [1, 2, '3', '3.4']]

lista_plana = aplanar(lista)
print(lista_plana)
```

```python
# Después la función para separar los datos de la nueva lista en flotantes o no

def organizador_listas(lista_plana):
    lista_flotantes = []
    lista_otros = []    

    for elemento in lista_plana:
        try: # usamos try-except para evitar que el programa falle con valores no numéricos
            num = float(elemento) # intentamos convertir a float
            lista_flotantes.append(num)  # si funciona: es un número flotante
        except (ValueError, TypeError): # valor incorrecto o tipo de dato incorrecto para la operación
            lista_otros.append(elemento)  # si falla: lo agregamos a la otra lista
    
    return lista_flotantes, lista_otros

lista_flotantes, lista_otros = organizador_listas(lista_plana)

print(f"lista_flotantes: {lista_flotantes}")
print(f"lista_otros: {lista_otros}")
```

```python
import re # importo el módulo re primero para poder utilizar la funcion re.sub()

with open("/Users/liss/Desktop/ItAcademy/2-Especialización/¡Sprints!/S7/Datos/tu_me_quieres_blanca.txt", "r") as texto2:
    texto2 = texto2.read() # leer el archivo de texto

texto2 = texto2.lower()  # Convertimos a minúsculas
texto2 = re.sub(r'[^\w\s]', '', texto2)  # Eliminamos signos de puntuación
texto2 = texto2.replace("\n", " ")  # Eliminamos saltos de línea

def contador_texto_ordenado(texto):
    dicc_texto = {}  
    lista_texto = sorted(texto.split())  # dividimos en palabras y ordenamos

    for palabra in lista_texto:
        letra_inicial = palabra[0]  # obtener la primera letra

        if letra_inicial not in dicc_texto:
            dicc_texto[letra_inicial] = {}  # crear la key si no existe

        if palabra in dicc_texto[letra_inicial]:
            dicc_texto[letra_inicial][palabra] += 1  # si la palabra ya está en el diccionario: aumentar el contador
        else:
            dicc_texto[letra_inicial][palabra] = 1  # si la palabra no está en el diccionario: inicializar el contador

    return dicc_texto

dicc_texto_ordenado = contador_texto_ordenado(texto2)

dicc_texto_ordenado
```
